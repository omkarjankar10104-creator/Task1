```python
'''1. Create an array A of shape (1000, 5) from a normal distribution with mean 50 and
standard deviation 12. Print its shape, dtype, and memory size in bytes. Convert it to
float32 and print the memory size again.'''

import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))

print("Shape:", A.shape)
print("Dtype:", A.dtype)
print("Memory size:", A.nbytes, "bytes")

A = A.astype(np.float32)

print("New dtype:", A.dtype)
print("Memory size after float32:", A.nbytes, "bytes")
```

```python
'''2. Standardise every column so each has mean 0 and standard deviation 1. Do it in one
expression using broadcasting. Print the column means and standard deviations after
scaling, rounded to 6 decimals'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))
A = A.astype(np.float32)

A = (A - A.mean(axis=0)) / A.std(axis=0)

print("Column means:")
print(np.round(A.mean(axis=0), 6))

print("Column standard deviations:")
print(np.round(A.std(axis=0), 6))
```

```python
'''3. Write minmax(X, axis) that scales to the range 0 to 1 along any axis. Test it on
axis=0 and axis=1 and show that the minimum and maximum come out as 0 and 1.
'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))


def minmax(X, axis):
    X_min = np.min(X, axis=axis, keepdims=True)
    X_max = np.max(X, axis=axis, keepdims=True)

    return (X - X_min) / (X_max - X_min)


# Axis = 0
result_0 = minmax(A, axis=0)

print("Axis 0 minimum:")
print(np.min(result_0, axis=0))

print("Axis 0 maximum:")
print(np.max(result_0, axis=0))


# Axis = 1
result_1 = minmax(A, axis=1)

print("\nAxis 1 minimum:")
print(np.min(result_1, axis=1))

print("Axis 1 maximum:")
print(np.max(result_1, axis=1))
```

```python
'''4. Inject NaN into 5% of the positions of A at random. Replace every NaN with the mean
of its own column, using masking and np.nanmean . Confirm with np.isnan(A).sum()
== 0 .'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))

# Number of NaN values = 5%
num_nan = int(A.size * 0.05)

# Random positions
indices = np.random.choice(A.size, num_nan, replace=False)

A.flat[indices] = np.nan

print("NaN values before replacement:", np.isnan(A).sum())


# Column means ignoring NaN
column_means = np.nanmean(A, axis=0)

# Find NaN positions
mask = np.isnan(A)

# Replace each NaN with its column mean
A[mask] = column_means[np.where(mask)[1]]

print("NaN values after replacement:", np.isnan(A).sum())

assert np.isnan(A).sum() == 0

print("Assertion passed.")
```

```python
'''5. Flag every value more than 3 standard deviations from its column mean and replace it
with the column median. Report how many values were replaced per column.'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))

# Column statistics
column_mean = np.mean(A, axis=0)
column_std = np.std(A, axis=0)

# Find values more than 3 standard deviations away
outlier_mask = np.abs(A - column_mean) > (3 * column_std)

# Count replacements per column
replaced = np.sum(outlier_mask, axis=0)

print("Values replaced per column:")
print(replaced)


# Column median
column_median = np.median(A, axis=0)

# Replace outliers with column median
A[outlier_mask] = column_median[np.where(outlier_mask)[1]]

print("Total values replaced:", np.sum(replaced))
```

```python
'''6. Take the first 200 rows and build the full 200 × 200 Euclidean distance matrix using
broadcasting only. Do not use loops, scipy , or sklearn . Prove it is correct by
checking that the diagonal is 0 and that the matrix equals its own transpose.
'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))

X = A[:200]

# Broadcasting
difference = X[:, np.newaxis, :] - X[np.newaxis, :, :]

# Euclidean distance
distance_matrix = np.sqrt(np.sum(difference ** 2, axis=2))

print("Distance matrix shape:", distance_matrix.shape)

# Diagonal should be zero
print(
    "Diagonal is zero:",
    np.allclose(np.diag(distance_matrix), 0)
)

# Matrix should equal its transpose
print(
    "Matrix equals transpose:",
    np.allclose(distance_matrix, distance_matrix.T)
)
```

```python
'''7. From that distance matrix, find for every row the indices of its 5 nearest rows, excluding
itself. Return the result as a (200, 5) integer array. Use argpartition , not a sort of
the whole row, and say in one line why.
'''
import numpy as np

np.random.seed(42)

A = np.random.normal(50, 12, (1000, 5))

X = A[:200]

# Euclidean distance matrix
difference = X[:, np.newaxis, :] - X[np.newaxis, :, :]
D = np.sqrt(np.sum(difference ** 2, axis=2))

# Exclude itself
np.fill_diagonal(D, np.inf)

# Get 5 nearest rows
nearest = np.argpartition(D, kth=4, axis=1)[:, :5]

print("Shape:", nearest.shape)
print("Dtype:", nearest.dtype)

print("\nFirst 5 rows:")
print(nearest[:5])
```

```python
'''8. Create a 1-D array of 1,000,000 random integers between 0 and 99. Count how many
times each value appears using np.bincount . Then count the same thing with a plain
Python loop. Put %%time at the top of each cell and report the speed-up as a number'''
import numpy as np

np.random.seed(42)

numbers = np.random.randint(0, 100, 1_000_000)

%time counts_numpy = np.bincount(numbers)

print(counts_numpy)
```

```python
import numpy as np

np.random.seed(42)

numbers = np.random.randint(0, 100, 1_000_000)

%time
counts_python = [0] * 100

for x in numbers:
    counts_python[x] += 1

print(counts_python)
```

```python
import numpy as np

print(
    "Same result:",
    np.array_equal(counts_numpy, np.array(counts_python))
)
```

```python
'''9. Build all sliding windows of length 10 over a 1-D array of 10,000 values using
sliding_window_view , imported with from numpy.lib.stride_tricks import
sliding_window_view . Print the resulting shape. Compute the mean of each window
and compare your answer against pandas.Series.rolling(10).mean() with
np.allclose .'''
import numpy as np
import pandas as pd
from numpy.lib.stride_tricks import sliding_window_view

np.random.seed(42)

x = np.random.random(10000)

# Create windows of length 10
windows = sliding_window_view(x, 10)

print("Window shape:", windows.shape)

# Mean of every window
window_mean = windows.mean(axis=1)

print("Mean shape:", window_mean.shape)


# Pandas comparison
pandas_mean = pd.Series(x).rolling(10).mean().to_numpy()

# Remove first 9 NaN values
pandas_mean = pandas_mean[9:]

print(
    "Answers match:",
    np.allclose(window_mean, pandas_mean)
)
```

```python
'''10. Reshape a (12, 8) array into (4, 3, 8) , then take the mean over the middle axis.
Explain in two lines what the resulting shape represents'''
import numpy as np

np.random.seed(42)

A = np.random.random((12, 8))

# Reshape
B = A.reshape(4, 3, 8)

print("Original shape:", A.shape)
print("Reshaped shape:", B.shape)

# Mean over middle axis
result = B.mean(axis=1)

print("Result shape:", result.shape)
```
